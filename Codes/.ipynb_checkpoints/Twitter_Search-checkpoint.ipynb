```python
"""
Collect tweets from Twitter containing the given search query.

Usage:
python <script-name>  <query>  <result type: recent / popular / mixed>  <N: number of tweets to collect>  <output file>
Crawled tweets (json objects) will be appended to the output file, along with the current timestamp (when the crawl was made) -- SG
"""

from __future__ import division, print_function

__author__  = "Parantapa Bhattacharya <pb@parantapa.net>"

import sys
import codecs
sys.stdout = codecs.getwriter('utf-8')(sys.stdout)

import json
import calendar

import time
from time import sleep

#import times
import requests
from requests_oauthlib import OAuth1

from logbook import Logger
log = Logger(__name__)

# Constants
RESET_BUFFER      = 5
RATE_LIMIT_BUFFER = 1
RETRY_AFTER       = 5
RETRY_MAX         = 500

def rest_rate_limit(r):
    """
    Check the rate limit and sleep it off if hit.
    """

    try:
        #limit  = int(r.headers["X-Rate-Limit-Limit"])
        remain = int(r.headers["X-Rate-Limit-Remaining"])
        reset  = int(r.headers["X-Rate-Limit-Reset"])

        # following two lines added -- SG
        # using the calendar and time libraries, instead of the times library, since times library is not a standard Debian library -- SG
        #print(r.headers["date"])	# Example value: "Tue, 03 Nov 2015 12:36:15 GMT"
        curtime = int(calendar.timegm(time.strptime(r.headers["date"], "%a, %d %b %Y %H:%M:%S GMT")))	

        # original version using the times library -- SG
        #curtime = times.to_unix(times.parse(r.headers["date"]))

    except KeyError as e:
        # We dont have the proper headers
        log.error("Header not found - {}", e)
        sleep(RETRY_AFTER)
        return

    if remain <= RATE_LIMIT_BUFFER:
        log.debug("Hit rate limit - {}", remain)
        log.debug("Rate limit reset in {} seconds", reset - curtime)
        sleep(reset - curtime + RESET_BUFFER)



def search_tweets(query, result_type, auth, maxtweets=sys.maxsize):
    """
    Get as many tweets from twitter search as possible.

    query - The twitter search query
    result_type - mixed, recent, popular

    Returns an iterable of tweets.
    """

    endpoint = 'https://api.twitter.com/1.1/search/tweets.json'
    auth = OAuth1(signature_type="auth_header", **auth)
    params = { "q": query, "result_type": result_type, "count": 100,
               "include_entities": "true", "since":"2016-03-05",  "until":"2016-03-10" }
    # lang: en implies we only want English tweets  -- SG

    # We gather all tweet ids here
    maxid = float("inf")
    tcount = 0

    tries = 0
    while tries < RETRY_MAX:

        r = requests.get(endpoint, params=params, auth=auth, timeout=60.0)
        #print("Status code: %d" % r.status_code)	# added for checking - SG

        # Proper receive
        if r.status_code == 200:
            data = r.json()

            num_tweets_fetched = len(data["statuses"])
            print("Number of tweets fetched: %d" % num_tweets_fetched)	# added for checking - SG
            if num_tweets_fetched == 0:	
                tries = RETRY_MAX	# if no tweets could be fetched, no point querying again. Force loop termination -- SG
            else:
                for tweet in data["statuses"]:
                    yield tweet
                    maxid = min(tweet["id"], maxid)
                    tcount += 1
                    if tcount >= maxtweets:
                        return

                # Set the new max_id value
                params["max_id"] = maxid - 1

                tries = 0
            # end else

            rest_rate_limit(r)
            continue

        # Check if rate limited
        if r.status_code == 429:
            log.info(u"Try {}: Being throttled - {} {}",
                     tries, r.status_code, r.text)
            rest_rate_limit(r)
            tries += 1
            continue

        # Dont expect anything else
        log.warn(u"Try {}: Unexpected response - {} {}",
                 tries, r.status_code, r.text)
        rest_rate_limit(r)
        tries += 1
        continue

    # end while

    # commenting out the next two lines, since we want to stop after RETRY_MAX tries even if the specified number of tweets have not been collected -- SG
    log.critical("Maximum retries exhausted ...")
    raise ValueError(query, result_type, auth, maxtweets)



def save(iterable, fobj):
    """
    Save the tweets.
    """

    count = 0
    for tweet in iterable:
        data   = json.dumps(tweet)

        fobj.write(data + "\n")
        count += 1

        # commenting out the next two lines, no need of output to screen -- SG
        # for checking - print out the username and the tweet
        #print(u"@{} - {}".format(tweet["user"]["screen_name"], tweet["text"]))
        
    #print("Written data of %d tweets" % count)	# line added for checking -- SG



def main():
    try:
        #_, query, result_type, n, ofname = sys.argv
        query="cricket"
        result_type="recent"
        n=400
        ofname="cricket.tweets.jsons"
    except ValueError:
        print(sys.argv[0], "<query> <result type: recent / popular / mixed>  <N: number of tweets to collect>  <output file>")
        sys.exit(1)


    if not(result_type == 'recent' or result_type == 'popular' or result_type == 'mixed'):
        print("Invalid result type.")
        sys.exit(1)

    """
    Valid result types (as specified in https://dev.twitter.com/rest/reference/get/search/tweets):
    recent: return only the most recent results in the response
    popular: return only the most popular results in the response
    mixed: include both popular and real time results in the response
    """

    current_time = time.strftime("%Y-%m-%d %H:%M:%S")	# line added -- SG

    # The purpose of float to int is to parse stuff like 1e9 = 1 Million
    n = int(float(n))

    # Twitter auth tokens

    
    # crawling account: atResh2016
    # password: _atResh2016_ 
    token = {
        "client_key"            : 'ZqgzZvw39hBHfigsyAKf0yLMq',
        "client_secret"         : 'T64rfwbZadGb95coR6wjbbt5iCdkAN48kIGkgHPG11E0VUMQU5',
        "resource_owner_key"    : '707849458864807938-BP5EZHRWoIi1H2Iob541bjifNNZE8rZ',
        "resource_owner_secret" : '0CYoPy6O6SpZe1Q1ewwe3Ntkz7a2CHk6bCWo31L0K8UiO',
    }
    

    # Get the tweets
    tweets = search_tweets(query, result_type, token, n)  

#     # Save them
#     with open(ofname, "a") as fobj:	# opening file in append mode -- SG
#         # write the current timestamp into the file -- SG
#         fobj.write("\n======== %s ========\n\n" % current_time)
#         save(tweets, fobj)

if __name__ == '__main__':
    main()
 

```
