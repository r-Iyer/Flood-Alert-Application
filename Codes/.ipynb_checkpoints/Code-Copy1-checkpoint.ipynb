```python
def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder,filename))
        if img is not None:
            images.append(img)
    return images
```

```python
images=load_images_from_folder("../InDir/sliding")
images2=load_images_from_folder("../InDir/slidingwater")
```

```python
images=np.asarray(images)
images2=np.asarray(images2)
im=np.concatenate((images,images2),axis=0)
```

```python
len(images)
```

```python
y=np.zeros(len(im))
y[len(images):]=1
```

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder
from keras.utils import np_utils
encoder = LabelEncoder()
encoder.fit(y)
y = encoder.transform(y)
y= np_utils.to_categorical(y)
```

```python
im[0].shape
```

```python
#filepath = "CNN-{epoch:02d}-{val_acc:.3f}"  # unique file name that will include the epoch and the validation acc for that epoch
#checkpoint = ModelCheckpoint("models/{}.model".format(filepath, monitor='val_acc', verbose=1, save_best_only=True, mode='max')) # saves only the best ones
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Activation, Flatten
from tensorflow.keras.layers import Conv2D, MaxPooling2D

model = Sequential()
model.add(Conv2D(64, (3,3), input_shape=(im[0].shape)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=2))

model.add(Conv2D(64, (3,3)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=2))

model.add(Flatten())  # this converts our 3D feature maps to 1D feature vectors

model.add(Dense(64))

model.add(Dense(y.shape[1]))
model.add(Activation('softmax'))
for layer in model.layers:
    
    print(layer.output_shape)
model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

```

```python
model.fit(im,y, batch_size=32, epochs=2, validation_split=0.3)
```

```python
model.save('../models/model1.h5')
```

```python
from keras.utils import CustomObjectScope
from keras.initializers import glorot_uniform
from keras.models import load_model
with CustomObjectScope({'GlorotUniform': glorot_uniform()}):
        model = load_model('../models/model1.h5')
```

```python
img1=cv2.imread('../test/image1.jpg',cv2.IMREAD_COLOR);
#img1 = cv2.resize(img1, (1024,1024), interpolation=cv2.INTER_AREA)
(winW, winH) = (48,48)
```

```python
img1.shape
```

```python
def sliding_window(image, stepSize, windowSize):
	# slide a window across the image
	for y in range(0, image.shape[0], stepSize):
		for x in range(0, image.shape[1], stepSize):
#			print(y + windowSize[1])
			# yield the current window
			yield (x, y, image[y:y + windowSize[1], x:x + windowSize[0]])
```

```python

for (x, y, window) in sliding_window(img1, stepSize=32, windowSize=(winW, winH)):
   #   print(window.shape[0]," ",window.shape[1])
    if window.shape[0] != winH or window.shape[1] != winW:
        continue
    print(window.shape)
    X= np.expand_dims(window, axis=0)
    Y_pred=model.predict(X)
    #img1[y:y + winW, x:x + winH,:]=Y_pred
    # since we do not have a classifier, we'll just draw the window
    clone = img1.copy()
    cv2.rectangle(clone, (x, y), (x + winW, y + winH), (0, 255,255*Y_pred[0]), 2)
   # list.append(clone)
    cv2.imshow("clone", clone)
    #cv2.imshow("window",window)
    cv2.waitKey(1)
    time.sleep(0.025)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
